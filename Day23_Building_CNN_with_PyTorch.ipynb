# Creating Convolution Layers using PyTorch

## **Let's build a 2D convolution layer**
- [nn.Conv2d()](https://pytorch.org/docs/stable/generated/torch.nn.Conv2d.html)
  - applies a 2D convolution over an input volume of $(C_{in}​,H_{in},W_{in})$ and produces an output volume of $(C_{out}​,H_{out},W_{out})$   between two adjacent layers.
  - to create this, you need to provide the followings:
    - __channel_dimension_of_input_layer__ i.e., $C_{in}$
    - __channel_dimension_of_output_layer__ i.e., $C_{out}$
    - __filter_size__ i.e., $F$

  - the other two optional parameters are __stride__: $S=1$ and __padding__: $P=0$, with default values as shown. As discussed in class, PyTorch will calculate internally the sizes of output volume $H_{out}$ and $W_{out}$ from the above mentioned parameter values


> For example, let's create a 2D convolution layer by specify the following:
>> input volume channels size is 1

>> output volume channel size is 32

>> each filter has a size of (3x3)

>> default stride size is 1

>> default padding size is 0

### **Create a convolution layer.**

```python
# import torch library
import torch
import numpy as np
import torch.nn as nn

conv_layer = nn.Conv2d( 8, 8, 3, 1, 0)
```

### **Create a random input volume of the given size.**

> More precisely, the input has 8 channels, a height of 6, and a width of 7. In other words, $(C_{in}=8, H_{in}=6, W_{in}=7)$.

```python
input_volume_tensor = torch.randn((1, 8, 6, 7))             # [HOW_MANY_TENSORS, CHANNEL_SIZE, HEIGHT, WIDTH]
print(f"input volume shape:", input_volume_tensor.shape)    # [HOW_MANY_TENSORS, CHANNEL_SIZE, HEIGHT, WIDTH]
```

### **Perform a forward pass of the input volume through the newly created convolutional layer.**

> More precisely, the input has 8 channels, a height of 6, and a width of 7. In other words, $(C_{in}=8, H_{in}=6, W_{in}=7)$.

```python
output_volume_tensor = conv_layer(input_volume_tensor)      # [HOW_MANY_TENSORS, CHANNEL_SIZE, HEIGHT, WIDTH]
print(f"output volume shape:", output_volume_tensor.shape)  # [HOW_MANY_TENSORS, CHANNEL_SIZE, HEIGHT, WIDTH]
```

###**Verify the shape of the output volume $(C_{out}​,H_{out},W_{out})$ by hand calculation**
There are 8 filters in this convolutional layer. Each filter has a size of (3x3), hence:

- $F=3$

You should use the default values of the other two optional parameters as follows (when they are not mentioned explicitly):
-  __stride__: $S=1$
- __padding__: $P=0$

Given the above parameters and our input volume of sizes $C_{in}$=8, $H_{in}$=6, $W_{in}$=6, we need to find/verify the following:

- $C_{out}$=?

- $H_{out}$=?

- $W_{out}$=?


To find the size of the output volume's height by hand calculation do the following:

> $H_{out}$ = ($H_{in}$ - $F$ + 2*$P$)/$S$ + 1

>> (6 - 3 + 2*0)/1 + 1

>> = 3/1 + 1

>> = 3 + 1

>> = 4

Hence output volume's height $H_{out}$ = 4

To find the size of the output volume's width by hand calculation do the following:

> $W_{out}$ = ($W_{in}$ - $F$ + 2*$P$)/$S$ + 1

>> (7 - 3 + 2*0)/1 + 1

>> = 4/1 + 1

>> = 4 + 1

>> = 5

Hence output volume's width $W_{out}$ = 5

## **Let's build a 2 layer convolutional neural network (CNN)!**

First convolution layer has:
  > its input volume has 3 channels (inputs are RGB color images)

  > the output volume will have 64 channels

  > each filter has a size of (3x3)

Second convolution layer has:
  > its input volume has 64 channels

  > the output volume will have 128 channels

  > each filter has a size of (3x3)


```python
# construction
input_volume_channel_first    = 3
output_volume_channel_first   = 64
input_volume_channel_second   = 64
output_volume_channel_second  = 128
filter_size                   = 3



simple_cnn1 = nn.Sequential(
  nn.Conv2d( 3, 64, 3, 1, 0),
  nn.ReLU(),
  nn.Conv2d(64, 128, 3, 1, 0),
  nn.ReLU()
)

```

```python
output = simple_cnn1(random_X)
output.shape
```
